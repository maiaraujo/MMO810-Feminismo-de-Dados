```python
def limpar(df):
    df = df.copy()
    df.columns = [str(c).strip() for c in df.columns]
    for c in df.columns:
        if not c.isdigit() and (df[c].dtype == 'object' or pd.api.types.is_string_dtype(df[c])):
            df[c] = df[c].str.strip()
    return df


def longo(df, ids):
    anos = [c for c in df.columns if c.isdigit()]
    t = df.melt(id_vars=ids, value_vars=anos, var_name='Ano', value_name='Valor')
    t = t.dropna(subset=['Valor']).copy()
    t['Ano'] = t['Ano'].astype(int)
    return t


raw = {k: limpar(v) for k, v in raw.items()}
SEXO = {'M': 'Masculino', 'F': 'Feminino'}
NAO_DECL = {'Não Declarado': 'Não declarado', 'Não Declarada': 'Não declarado'}

# --- Alunos
al_gen = longo(raw['Alunos - Gênero'], ['Unidade', 'Gênero'])
al_gen['Sexo'] = al_gen['Gênero'].map(SEXO)

al_raca = longo(raw['Alunos - RaçaCor'], ['Unidade', 'Raça/Cor'])
al_raca['Raça/Cor'] = al_raca['Raça/Cor'].replace(NAO_DECL)
al_raca['Percentual'] = al_raca['Valor'] * 100          # fração -> %

al_pcd = longo(raw['Alunos - PCD'], ['Unidade', 'Categoria / Condição'])

# --- Docentes
doc_gen = longo(raw['Docentes - Gênero'], ['Unidade', 'Regime de Trabalho', 'Nível / Carreira', 'Sexo'])
doc_gen['Sexo'] = doc_gen['Sexo'].map(SEXO)

doc_raca = longo(raw['Docentes - RaçaCor'], ['Unidade', 'Raça/Cor'])
doc_raca['Raça/Cor'] = doc_raca['Raça/Cor'].replace(NAO_DECL)

doc_pcd = longo(raw['Docentes - PCD'], ['Unidade'])

# --- Apoio
UNIDADES = list(raw['Alunos - Gênero']['Unidade'].unique())
TOTAL_MS = 'Total MS (Todos os Regimes)'
# docentes MS por unidade (todos os regimes somados)
DOC_UN = doc_gen[(doc_gen['Regime de Trabalho'] == TOTAL_MS) & (doc_gen['Unidade'].isin(UNIDADES))]
# série histórica UNICAMP (Carreira MS)
DOC_TOTAL = doc_gen[(doc_gen['Regime de Trabalho'] == TOTAL_MS) & (doc_gen['Unidade'] == 'UNICAMP (Total MS)')]


def tabela_sexo(df, ano):
    """Feminino, Masculino, Total e % Feminino por unidade em um ano."""
    t = df[(df['Ano'] == ano) & (df['Unidade'].isin(UNIDADES))]
    p = t.pivot_table(index='Unidade', columns='Sexo', values='Valor', aggfunc='sum').fillna(0)
    p['Total'] = p['Feminino'] + p['Masculino']
    p['% Feminino'] = p['Feminino'] / p['Total'] * 100
    return p

# FALLBACK: a planilha não traz a linha 'UNICAMP (Total MS)' -> o código soma as 24 unidades
if DOC_TOTAL.empty:
    DOC_TOTAL = DOC_UN.groupby(['Ano', 'Sexo'], as_index=False)['Valor'].sum()
    DOC_TOTAL['Unidade'] = 'UNICAMP (Total MS)'
```

## 7. Preparação — tabela única por instituto (2025)

```python
ANO = 2025
PPI = ['Preta', 'Parda', 'Indígena']

# --- alunos
_g = tabela_sexo(al_gen, ANO)
PAN = pd.DataFrame({'Alunos': _g['Total'], '% Alunas': _g['% Feminino']})

# --- docentes: gênero (Carreira MS, todos os regimes)
_d = DOC_UN[DOC_UN['Ano'] == ANO].pivot_table(index='Unidade', columns='Sexo', values='Valor', aggfunc='sum').fillna(0)
PAN['Docentes'] = _d['Feminino'] + _d['Masculino']
PAN['% Docentes mulheres'] = _d['Feminino'] / PAN['Docentes'] * 100

# --- docentes: raça/cor (docentes ativos)
_r = doc_raca[(doc_raca['Ano'] == ANO) & doc_raca['Unidade'].isin(UNIDADES)] \
        .pivot_table(index='Unidade', columns='Raça/Cor', values='Valor', aggfunc='sum').fillna(0)
_tot = _r.sum(axis=1)
PAN['% Docentes PPI'] = _r[PPI].sum(axis=1) / _tot * 100
PAN['% Docentes brancos'] = _r['Branca'] / _tot * 100
PAN['% Docentes não declarado'] = _r['Não declarado'] / _tot * 100

# --- PCD
_p = al_pcd[(al_pcd['Ano'] == ANO) & al_pcd['Unidade'].isin(UNIDADES)].set_index('Unidade')['Valor']
PAN['Alunos PCD'] = _p
PAN['PCD alunos por 1.000'] = PAN['Alunos PCD'] / PAN['Alunos'] * 1000
_f = doc_pcd[(doc_pcd['Ano'] == ANO)].set_index('Unidade')['Valor']
PAN['Func. PCD'] = _f
PAN['Gap mulheres (docentes − alunas, p.p.)'] = PAN['% Docentes mulheres'] - PAN['% Alunas']
PAN = PAN.loc[UNIDADES]
PAN.round(1)
```

### Gráfico G — "Se a UNICAMP tivesse 100 pessoas": alunos × docentes por raça/cor
Cada ponto = 1% do grupo. Compara o perfil racial de quem estuda com o de quem ensina.

```python
from plotly.subplots import make_subplots
_ar = al_raca[al_raca['Ano'] == ANO].set_index('Raça/Cor')['Percentual'].reindex(ORDEM_RACA)
_dr = dr[dr['Unidade'].isin(UNIDADES)].groupby('Raça/Cor')['Valor'].sum().reindex(ORDEM_RACA)
_dr = _dr / _dr.sum() * 100

def fonte(fig, x=16, t=24):
    fig.update_layout(font=dict(size=x), title_font_size=t, legend_font_size=x + 2)
    return fig

def waffle(serie, ordem, cores):
    n = np.round(serie.reindex(ordem).fillna(0)).astype(int)
    n.iloc[-1] += 100 - n.sum()                      # ajusta arredondamento
    xs, ys, cs, ns = [], [], [], []
    i = 0
    for cat, k in n.items():
        for _ in range(max(k, 0)):
            xs.append(i % 10); ys.append(i // 10); cs.append(cores[cat]); ns.append(cat); i += 1
    return xs, ys, cs, ns

fig = make_subplots(rows=1, cols=2, subplot_titles=['100 alunos', '100 docentes'])
for j, s in enumerate([_ar, _dr], start=1):
    xs, ys, cs, ns = waffle(s, ORDEM_RACA, COR_RACA)
    fig.add_trace(go.Scatter(x=xs, y=ys, mode='markers', marker=dict(size=26, color=cs), text=ns,
                             hovertemplate='%{text}<extra></extra>', showlegend=False), 1, j)
for cat in ORDEM_RACA:                                  # legenda
    fig.add_trace(go.Scatter(x=[None], y=[None], mode='markers', marker=dict(size=14, color=COR_RACA[cat]), name=cat))
fig.update_xaxes(visible=False); fig.update_yaxes(visible=False, autorange='reversed')
fig.update_layout(title='Se a UNICAMP tivesse 100 pessoas: alunos × docentes por raça/cor (2025)')
mostrar(fonte(fig), 'criativo_waffle_raca', 600)
```

```python
# Números absolutos do Excel
_a = tabela_sexo(al_gen, ANO).loc[UN]
a_f, a_m = int(_a['Feminino']), int(_a['Masculino'])

_d = DOC_UN[
    (DOC_UN['Ano'] == ANO) & (DOC_UN['Unidade'] == UN)
].groupby('Sexo')['Valor'].sum()

d_f, d_m = int(_d['Feminino']), int(_d['Masculino'])

def pontos(f, m):
    """100 pontos: mulheres primeiro, depois homens."""
    nf = round(f / (f + m) * 100)
    sexos = ['Feminino'] * nf + ['Masculino'] * (100 - nf)

    xs = [i % 10 for i in range(100)]
    ys = [i // 10 for i in range(100)]

    simbolos = [
        'triangle-up' if s == 'Feminino' else 'triangle-down'
        for s in sexos
    ]

    return xs, ys, [COR_SEXO[s] for s in sexos], sexos, simbolos


fig = make_subplots(
    rows=1, cols=2, horizontal_spacing=0.08,
    subplot_titles=[
        f'Alunos do {UN}: {a_f} F | {a_m} M  ({a_f/(a_f+a_m)*100:.1f}% mulheres)',
        f'Docentes do {UN}: {d_f} F | {d_m} M  ({d_f/(d_f+d_m)*100:.1f}% mulheres)'
    ]
)

for j, (f, m) in enumerate([(a_f, a_m), (d_f, d_m)], start=1):
    xs, ys, cs, ns, simbolos = pontos(f, m)

    fig.add_trace(
        go.Scatter(
            x=xs,
            y=ys,
            mode='markers',
            marker=dict(
                size=22,
                color=cs,
                symbol=simbolos
            ),
            text=ns,
            hovertemplate='%{text}<extra></extra>',
            showlegend=False
        ),
        1, j
    )

# Legenda com cor e geometria
for sx, simbolo in [
    ('Feminino', 'triangle-up'),
    ('Masculino', 'triangle-down')
]:
    fig.add_trace(
        go.Scatter(
            x=[None], y=[None],
            mode='markers',
            marker=dict(
                size=14,
                color=COR_SEXO[sx],
                symbol=simbolo
            ),
            name=sx
        )
    )

fig.update_xaxes(visible=False)
fig.update_yaxes(visible=False, autorange='reversed')

fig.update_layout(
    title=f'Se o {UN} tivesse 100 pessoas: mulheres entre alunos e docentes (2025)'
)

mostrar(fonte(fig, 14, 20), f'criativo_waffle_genero_{UN}', 480)
```
